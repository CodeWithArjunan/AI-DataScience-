# What is LangChain?
LangChain is a framework built to facilitate the creation of applications powered by large language models (LLMs). It abstracts away many of the complexities involved in managing language models, from prompt engineering to chaining different language models together. LangChain supports different LLM providers like OpenAI, Hugging Face, and Cohere, making it a versatile choice for developers.

# Key use cases of LangChain include:

 * Question Answering (QA) systems
 * Text generation and Summarization
 * Conversational agents (chatbots)
 * Automated Document Processing
 * Sentiment Analysis and Opinion Mining
 * and many more ….


#  Components of LLM Prompt templates, chains & LLMs
Basic Components of LangChain
Core Concepts in LangChain
LangChain simplifies working with LLMs by organizing tasks into several components. Understanding these components is essential to building any application using the framework. Below are the key building blocks:

# 1. Prompts/Prompt Templates
Prompts are at the heart of interacting with language models. LangChain provides a modular prompt system, allowing developers to dynamically build and modify prompts. You can create templates that combine variables, making it easier to scale or reuse your prompts across different models.

Note: Prompts with variables are known as prompt templates

Example: Below is the summarisation prompt template, that takes an paragraph as input and returns a short summary with neutral tone.

summary_prompt = '''

Instruction: 
Summarize the following text into a concise and coherent version. 
Focus on the key points and eliminate unnecessary details. 
Ensure the summary is significantly shorter than the original 
content while retaining the main message.

Tone & Style:
neutral

Length of Summary:
100 words

Input:
{input}
'''

# 2. LLMs (Large Language Models)
LangChain is designed to integrate with multiple LLMs. You can use APIs from models like OpenAI’s GPT-4, Hugging Face transformers, or any custom LLMs. LangChain makes it easier to switch between models without rewriting large parts of your application.

Lang chain can also connect to locally deployed LLM models through Ollama or vllm.These platform offers optimised infrastructure for fast, scalable, and cost-efficient language model deployments. For use-cases with private dataset / less infrastructure / education purposes, these kind of local deployed LLMs are the best solutions.


# 3. Chains
A chain is the central building block in LangChain. It allows you to combine multiple actions or language model calls in a sequence to achieve more complex tasks. For example, you can have a chain that first generates a summary and then performs sentiment analysis on that summary.

Write on Medium
Example:

chain = prompTemplate | model
Chains are very easy to be use in LangChain as simple as just using a pipe “|” operator. It can connect multiple langchain components.


# 4. Tools
LangChain provides various pre-built tools that can extend the capabilities of your application. For example, LangChain includes built-in support for external APIs, search engines, and databases. This enables you to chain language model queries with other processes like fetching data from an API or performing a web search.


# Key tools available in LangChain, each with a brief explanation:


 * WebSearchTool: Allows models to search the web for real-time information.
 * PythonRequestsTool: Enables models to send HTTP requests to external APIs for fetching live data.
 * Calculator Tool: Performs arithmetic and advanced mathematical operations for precise calculations.
 * Python REPL Tool: Executes Python code for dynamic calculations or script-based tasks.
 * Database Tool: Connects and queries databases to retrieve or manipulate structured data.
 * Wikipedia Tool: Fetches relevant information directly from Wikipedia for knowledge retrieval.
 * Code Interpreter Tool: Allows models to write and execute code, useful for solving coding or logic problems.
 * CSV Tool: Reads and processes CSV files, useful for extracting data or generating insights from tabular data.
 * Zapier Tool: Automates workflows by connecting models to over 5,000 apps via the Zapier platform.
 * Google Search Tool: Queries Google Search to retrieve the latest or specific information from the web.
 * Selenium Browser Tool: Automates browser interactions, such as scraping web data or testing UI components.
 * SerpAPI Tool: Provides Google search results using the SerpAPI, returning top search results for queries.

 
These tools enable LangChain to interact with external data, APIs, and perform tasks beyond the typical capabilities of language models.

# LangChain vs. LangGraph vs. Deep Agents
If you are looking to build an agent, we recommend you start with Deep Agents which comes “batteries-included”, with modern features like automatic compression of long conversations, a virtual filesystem, and subagent-spawning for managing and isolating context.

Deep Agents are implementations of LangChain agents. If you don’t need these capabilities or would like to customize your own for your agents and autonomous applications, start with LangChain.

Use LangGraph, our low-level agent orchestration framework and runtime, when you have more advanced needs that require a combination of deterministic and agentic workflows and heavy customization.

LangChain agents are built on top of LangGraph in order to provide durable execution, streaming, human-in-the-loop, persistence, and more. You do not need to know LangGraph for basic LangChain agent usage.
We recommend you use LangChain if you want to quickly build agents and autonomous applications.

```python
# pip install -qU langchain "langchain[anthropic]"
from langchain.agents import create_agent

def get_weather(city: str) -> str:
    """Get weather for a given city."""
    return f"It's always sunny in {city}!"

agent = create_agent(
    model="claude-sonnet-4-6",
    tools=[get_weather],
    system_prompt="You are a helpful assistant",
)


# Run the agent
agent.invoke(
    {"messages": [{"role": "user", "content": "what is the weather in sf"}]}
)
```
